```python
import os
from pathlib import Path
from dotenv import load_dotenv
from langchain_community.tools.tavily_search import TavilySearchResults

# 1. Load the .env file explicitly
current_dir = Path.cwd()
env_path = current_dir / ".env" if (current_dir / ".env").exists() else current_dir.parent / ".env"
load_dotenv(dotenv_path=env_path)

# 2. Initialize the search tool
search_tool = TavilySearchResults(
    max_results=3,
    tavily_api_key=os.getenv("TAVILY_API_KEY")
)

# 3. Test the search tool
result = search_tool.invoke({"query": "give me latest news on innovation in Fintech?"})
print(result)
```

```python
import os
from pathlib import Path
from dotenv import load_dotenv
from langchain_groq import ChatGroq

# Find and load .env from current directory or parent directory
current_dir = Path.cwd()
env_path = current_dir / ".env" if (current_dir / ".env").exists() else current_dir.parent / ".env"
load_dotenv(dotenv_path=env_path)

groq_key = os.getenv("GROQ_API_KEY")
print("GROQ_API_KEY loaded:", bool(groq_key))

llm = ChatGroq(
    model_name="llama-3.3-70b-versatile",
    temperature=0,
    groq_api_key=groq_key
)
```
